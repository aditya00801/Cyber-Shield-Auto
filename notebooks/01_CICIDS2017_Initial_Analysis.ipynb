```python
for file in sorted(DATA_DIR.glob("*.csv")):

    temp = pd.read_csv(file)
    temp.columns = temp.columns.str.strip()

    missing = temp.isnull().sum()

    print("=" * 70)
    print(file.name)

    if missing.sum() == 0:
        print("No missing values")
    else:
        print(missing[missing > 0])
```

```python
# infinite valuesd

for file in sorted(DATA_DIR.glob("*.csv")):

    temp = pd.read_csv(file)
    temp.columns = temp.columns.str.strip()

    numeric = temp.select_dtypes(include=np.number)

    inf_count = np.isinf(numeric).sum().sum()

    print(f"{file.name}: {inf_count} infinite values")
```

## Conclusion

The initial analysis of the CICIDS2017 dataset provided a clear understanding of its structure, traffic classes, and data-quality characteristics.

The dataset contains multiple types of network traffic, including BENIGN traffic and several attack categories such as DDoS, PortScan, Bot, Infiltration, Web Attacks, Patator attacks, DoS attacks, and Heartbleed. The analysis also showed significant class imbalance, with some attack classes containing a large number of samples while others are extremely rare.

Several data-quality issues were identified. Missing values were found in the `Flow Bytes/s` feature, and infinite values were present in multiple files. A considerable number of duplicate rows were also observed. These findings indicate that preprocessing is required before the dataset can be used for machine-learning model development.

The analysis also showed that the original dataset should be preserved and that preprocessing should be performed separately without modifying the raw CSV files. Duplicate handling, missing-value treatment, infinite-value handling, and a leakage-resistant train/test strategy must be carefully designed.

Based on these findings, the CICIDS2017 dataset is suitable for the next stage of CyberShield AI development. The next phase will focus on preprocessing, feature preparation, and the development of a multiclass intrusion detection model.

### Key Takeaways

- The dataset contains both normal and multiple attack categories.
- The dataset has significant class imbalance.
- `Flow Bytes/s` contains missing values.
- Infinite values occur in several flow-derived features.
- Duplicate rows are present and require careful treatment.
- Raw dataset files should remain unchanged.
- Data leakage must be considered when designing the train/test strategy.
- Further preprocessing is required before model training.
